### SEGUNDA PARTE
INNER JOIN PARA TRABAJAR CON AMBAS COLUMNAS

```python
conn.execute("""CREATE VIEW agencia_viviendas AS
SELECT v.*, a.AGENCY_NAME, a.RATING 
FROM viviendas AS v
INNER JOIN agencias AS a 
ON v.AGENCY_ID = a.AGENCY_ID;""")
```
